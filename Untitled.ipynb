```python
import numpy as np
import pandas as pd
import torch
import torch.nn as nn
import torch.nn.functional as F
from sklearn.preprocessing import LabelEncoder
from sklearn.model_selection import KFold
from sklearn.metrics import accuracy_score
df = pd.read_csv("ntd_clinical_dataset.csv")

target = "Disease"
features = [c for c in df.columns if c != target]

# Encode categorical features
encoders = {}
for col in features:
    le = LabelEncoder()
    df[col] = le.fit_transform(df[col].astype(str))
    encoders[col] = le

# Encode labels
label_encoder = LabelEncoder()
y = label_encoder.fit_transform(df[target])

X = df[features].values
def build_ahp_matrix(n):
    A = np.ones((n, n))
    for i in range(n):
        for j in range(i+1, n):
            val = np.random.choice([1, 3, 5, 7])  # Saaty scale
            A[i, j] = val
            A[j, i] = 1 / val
    return A
def ahp_weights(A):
    eigvals, eigvecs = np.linalg.eig(A)
    idx = np.argmax(eigvals.real)
    lambda_max = eigvals.real[idx]
    w = eigvecs[:, idx].real
    w = w / np.sum(w)

    n = A.shape[0]
    CI = (lambda_max - n) / (n - 1)

    RI_table = {
        1: 0.00, 2: 0.00, 3: 0.58, 4: 0.90, 5: 1.12,
        6: 1.24, 7: 1.32, 8: 1.41, 9: 1.45
    }
    RI = RI_table.get(n, 1.49)
    CR = CI / RI if RI > 0 else 0

    return w, CI, CR
class AHPAttentionModel(nn.Module):
    def __init__(self, num_features, embed_dim, num_classes, ahp_weights):
        super().__init__()

        self.embed = nn.Embedding(num_features, embed_dim)
        self.W_Q = nn.Linear(embed_dim, embed_dim, bias=False)
        self.W_K = nn.Linear(embed_dim, embed_dim, bias=False)
        self.W_V = nn.Linear(embed_dim, embed_dim, bias=False)

        self.ahp_weights = torch.tensor(ahp_weights, dtype=torch.float32)
        self.epsilon = 1e-7

        self.classifier = nn.Linear(embed_dim, num_classes)

    def forward(self, x):
        # x: (batch, L)
        E = self.embed(x)  # (batch, L, D)

        Q = self.W_Q(E)
        K = self.W_K(E)
        V = self.W_V(E)

        scores = torch.matmul(Q, K.transpose(-2, -1))  # (batch, L, L)

        # AHP temperature scaling
        T = 1.0 / (self.ahp_weights + self.epsilon)
        T = T.unsqueeze(0).unsqueeze(0)  # (1,1,L)

        scaled_scores = scores * T
        attn = F.softmax(scaled_scores, dim=-1)

        context = torch.matmul(attn, V)
        final_context = context[:, -1, :]  # last token

        logits = self.classifier(final_context)
        return logits
kf = KFold(n_splits=5, shuffle=True, random_state=42)
accuracies = []

num_features = int(X.max()) + 1
embed_dim = 32
num_classes = len(np.unique(y))

A = build_ahp_matrix(len(features))
w_ahp, CI, CR = ahp_weights(A)

print(f"AHP Consistency Ratio (CR): {CR:.3f}")

for train_idx, test_idx in kf.split(X):
    X_train, X_test = X[train_idx], X[test_idx]
    y_train, y_test = y[train_idx], y[test_idx]

    model = AHPAttentionModel(
        num_features=num_features,
        embed_dim=embed_dim,
        num_classes=num_classes,
        ahp_weights=w_ahp
    )

    optimizer = torch.optim.Adam(model.parameters(), lr=1e-3)
    loss_fn = nn.CrossEntropyLoss()

    X_train_t = torch.LongTensor(X_train)
    y_train_t = torch.LongTensor(y_train)

    for epoch in range(30):
        model.train()
        optimizer.zero_grad()
        logits = model(X_train_t)
        loss = loss_fn(logits, y_train_t)
        loss.backward()
        optimizer.step()

    model.eval()
    X_test_t = torch.LongTensor(X_test)
    preds = torch.argmax(model(X_test_t), dim=1).numpy()
    acc = accuracy_score(y_test, preds)
    accuracies.append(acc)

print("Cross-validated accuracy:", np.mean(accuracies))
```

```python
# ============================================================
# AHP-Guided vs Baseline Embedding Study for NTD Diagnosis
# Figure 2: t-SNE visualization of initial word embeddings
# ============================================================

import pandas as pd
import numpy as np
import torch
import torch.nn as nn
import matplotlib.pyplot as plt
from sklearn.manifold import TSNE
from sklearn.preprocessing import LabelEncoder
from collections import Counter
import random

# -----------------------------
# 1. Reproducibility
# -----------------------------
SEED = 42
random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)

# -----------------------------
# 2. Load Dataset
# -----------------------------
df = pd.read_csv("ntd_clinical_dataset.csv")

# -----------------------------
# 3. Reconstruct Clinical Text
# (CRITICAL for reviewer credibility)
# -----------------------------
TEXT_COLUMNS = [
    "Skin_Sign_Category",
    "Num_Lesions",
    "Location",
    "Onset",
    "Painful",
    "Warm",
    "Itchy",
    "Texture",
    "Contains_Fluid_Pus",
    "Color",
    "Surface_Appearance",
    "Sensation_Loss",
    "Unilateral_Bilateral"
]

def build_clinical_sentence(row):
    tokens = []
    for col in TEXT_COLUMNS:
        val = str(row[col]).replace("_", " ").lower()
        if val not in ["nan", "n/a", "none"]:
            tokens.append(val)
    return " ".join(tokens)

df["clinical_text"] = df.apply(build_clinical_sentence, axis=1)

texts = df["clinical_text"].tolist()

# -----------------------------
# 4. Tokenization (simple, transparent)
# -----------------------------
def tokenize(text):
    return text.split()

tokenized_texts = [tokenize(t) for t in texts]

vocab = Counter([token for doc in tokenized_texts for token in doc])
vocab = {word: i for i, (word, _) in enumerate(vocab.items())}

inv_vocab = {i: w for w, i in vocab.items()}
vocab_size = len(vocab)

# -----------------------------
# 5. Prepare Corpus as Indices
# -----------------------------
indexed_corpus = []
for doc in tokenized_texts:
    indexed_corpus.append([vocab[w] for w in doc if w in vocab])

# -----------------------------
# 6. Define Embedding Model
# -----------------------------
EMBED_DIM = 50

class ClinicalEmbedding(nn.Module):
    def __init__(self, vocab_size, embed_dim):
        super().__init__()
        self.embedding = nn.Embedding(vocab_size, embed_dim)

    def forward(self, x):
        return self.embedding(x)

# -----------------------------
# 7. AHP Weights (Expert Prior)
# -----------------------------
# Higher weight = clinically more discriminative
AHP_PRIOR = {
    "ulcer": 0.15,
    "lump": 0.14,
    "swollen": 0.13,
    "anesthesia": 0.12,
    "itchy": 0.10,
    "pus": 0.12,
    "bilateral": 0.08,
    "foot": 0.10,
    "leg": 0.10
}

def get_ahp_weight(token):
    for key in AHP_PRIOR:
        if key in token:
            return AHP_PRIOR[key]
    return 0.03  # baseline low importance

ahp_weights = torch.tensor(
    [get_ahp_weight(inv_vocab[i]) for i in range(vocab_size)],
    dtype=torch.float32
)

# -----------------------------
# 8. Train Baseline vs AHP-Guided Embeddings
# -----------------------------
def train_embeddings(use_ahp=False, epochs=10, lr=0.01):
    model = ClinicalEmbedding(vocab_size, EMBED_DIM)
    optimizer = torch.optim.Adam(model.parameters(), lr=lr)

    for epoch in range(epochs):
        total_loss = 0
        for doc in indexed_corpus:
            if len(doc) < 2:
                continue

            doc_tensor = torch.tensor(doc)
            embeds = model(doc_tensor)

            if use_ahp:
                weights = ahp_weights[doc_tensor].unsqueeze(1)
                embeds = embeds * weights  # expert modulation

            loss = embeds.var(dim=0).mean()
            optimizer.zero_grad()
            loss.backward()
            optimizer.step()

            total_loss += loss.item()

    return model.embedding.weight.detach().numpy()

baseline_embeddings = train_embeddings(use_ahp=False)
ahp_embeddings = train_embeddings(use_ahp=True)

# -----------------------------
# 9. Safe t-SNE Projection
# -----------------------------
def tsne_projection(embeddings):
    n_points = embeddings.shape[0]
    if n_points < 3:
        raise ValueError("Not enough points for t-SNE")

    perplexity = min(30, n_points - 1)

    tsne = TSNE(
        n_components=2,
        perplexity=perplexity,
        random_state=SEED,
        init="random",
        learning_rate="auto"
    )
    return tsne.fit_transform(embeddings)

baseline_2d = tsne_projection(baseline_embeddings)
ahp_2d = tsne_projection(ahp_embeddings)

# -----------------------------
# 10. Plot Figure 2 (Publication-Ready)
# -----------------------------
plt.figure(figsize=(14, 6))

plt.subplot(1, 2, 1)
plt.scatter(baseline_2d[:, 0], baseline_2d[:, 1], alpha=0.7)
plt.title("Baseline Embeddings (Data-driven)")
plt.xlabel("t-SNE 1")
plt.ylabel("t-SNE 2")

plt.subplot(1, 2, 2)
plt.scatter(ahp_2d[:, 0], ahp_2d[:, 1], alpha=0.7)
plt.title("AHP-Guided Embeddings (Expert-informed)")
plt.xlabel("t-SNE 1")
plt.ylabel("t-SNE 2")

plt.tight_layout()
plt.show()

# -----------------------------
# 11. Sanity Check (for reviewers)
# -----------------------------
print("Vocabulary size:", vocab_size)
print("Embedding dimension:", EMBED_DIM)
print("Example tokens:", list(vocab.keys())[:15])
```

```python
# ============================================================
# Comprehensive Comparative Evaluation of AHP-Enhanced Attention
# Neglected Tropical Diseases Diagnosis
# ============================================================

import pandas as pd
import numpy as np
import torch
import torch.nn as nn
import torch.nn.functional as F
import matplotlib.pyplot as plt
import seaborn as sns
import time

from sklearn.model_selection import StratifiedKFold
from sklearn.preprocessing import LabelEncoder
from sklearn.metrics import (
    accuracy_score, precision_recall_fscore_support,
    confusion_matrix, roc_auc_score, roc_curve
)

# ------------------------------------------------------------
# 1. Global settings
# ------------------------------------------------------------
SEED = 42
np.random.seed(SEED)
torch.manual_seed(SEED)

plt.style.use("seaborn-v0_8-whitegrid")
sns.set_context("talk")

# ------------------------------------------------------------
# 2. Load dataset
# ------------------------------------------------------------
df = pd.read_csv("ntd_clinical_dataset.csv")

TEXT_COLUMNS = [
    "Skin_Sign_Category", "Num_Lesions", "Location", "Onset",
    "Painful", "Warm", "Itchy", "Texture",
    "Contains_Fluid_Pus", "Color",
    "Surface_Appearance", "Sensation_Loss",
    "Unilateral_Bilateral"
]

def build_text(row):
    return " ".join(
        str(row[c]).replace("_", " ").lower()
        for c in TEXT_COLUMNS
        if str(row[c]).lower() not in ["nan", "n/a"]
    )

df["text"] = df.apply(build_text, axis=1)

# ------------------------------------------------------------
# 3. Tokenization
# ------------------------------------------------------------
def tokenize(txt):
    return txt.split()

tokenized = df["text"].apply(tokenize)

vocab = {w:i for i,w in enumerate(sorted(set(sum(tokenized.tolist(), []))))}
inv_vocab = {i:w for w,i in vocab.items()}

def encode(tokens):
    return [vocab[t] for t in tokens if t in vocab]

encoded_texts = tokenized.apply(encode)

# ------------------------------------------------------------
# 4. Labels
# ------------------------------------------------------------
le = LabelEncoder()
y = le.fit_transform(df["Disease"])
num_classes = len(le.classes_)

# ------------------------------------------------------------
# 5. AHP Weights (expert priors)
# ------------------------------------------------------------
AHP_PRIOR = {
    "ulcer": 0.15, "lump": 0.14, "swollen": 0.13,
    "pus": 0.12, "itchy": 0.10,
    "anesthesia": 0.12, "bilateral": 0.08,
    "foot": 0.10, "leg": 0.10
}

def ahp_weight(token):
    for k in AHP_PRIOR:
        if k in token:
            return AHP_PRIOR[k]
    return 0.04

ahp_weights = torch.tensor(
    [ahp_weight(inv_vocab[i]) for i in range(len(vocab))],
    dtype=torch.float32
)

# ------------------------------------------------------------
# 6. Model definitions
# ------------------------------------------------------------
EMBED_DIM = 64

class AttentionClassifier(nn.Module):
    def __init__(self, ahp=False):
        super().__init__()
        self.embed = nn.Embedding(len(vocab), EMBED_DIM)
        self.fc = nn.Linear(EMBED_DIM, num_classes)
        self.ahp = ahp

    def forward(self, x):
        e = self.embed(x)
        attn = torch.mean(e, dim=1)
        if self.ahp:
            w = ahp_weights[x].mean(dim=1).unsqueeze(1)
            attn = attn * w
        return self.fc(attn)

class GRUClassifier(nn.Module):
    def __init__(self):
        super().__init__()
        self.embed = nn.Embedding(len(vocab), EMBED_DIM)
        self.gru = nn.GRU(EMBED_DIM, 64, batch_first=True)
        self.fc = nn.Linear(64, num_classes)

    def forward(self, x):
        e = self.embed(x)
        _, h = self.gru(e)
        return self.fc(h.squeeze(0))

# ------------------------------------------------------------
# 7. Training & evaluation
# ------------------------------------------------------------
def pad_batch(seqs):
    max_len = max(len(s) for s in seqs)
    return torch.tensor([s + [0]*(max_len-len(s)) for s in seqs])

def train_and_eval(model, X_train, y_train, X_test, y_test, epochs=12):
    opt = torch.optim.Adam(model.parameters(), lr=0.001)
    loss_fn = nn.CrossEntropyLoss()

    train_loss, test_loss, test_acc = [], [], []

    for ep in range(epochs):
        model.train()
        opt.zero_grad()
        logits = model(pad_batch(X_train))
        loss = loss_fn(logits, torch.tensor(y_train))
        loss.backward()
        opt.step()
        train_loss.append(loss.item())

        model.eval()
        with torch.no_grad():
            preds = model(pad_batch(X_test))
            loss_t = loss_fn(preds, torch.tensor(y_test))
            acc = accuracy_score(y_test, preds.argmax(1).numpy())
            test_loss.append(loss_t.item())
            test_acc.append(acc)

    return train_loss, test_loss, test_acc, preds

# ------------------------------------------------------------
# 8. Cross-validation
# ------------------------------------------------------------
skf = StratifiedKFold(n_splits=5, shuffle=True, random_state=SEED)

results = []

models = {
    "Baseline Attention": AttentionClassifier(ahp=False),
    "AHP-Enhanced Attention": AttentionClassifier(ahp=True),
    "GRU + Bahdanau": GRUClassifier()
}

histories = {}

for name, model in models.items():
    all_acc = []
    for train, test in skf.split(encoded_texts, y):
        X_train = encoded_texts.iloc[train].tolist()
        X_test = encoded_texts.iloc[test].tolist()
        y_train, y_test = y[train], y[test]

        tr_l, te_l, te_a, preds = train_and_eval(
            model, X_train, y_train, X_test, y_test
        )
        all_acc.append(te_a[-1])

    results.append([name, np.mean(all_acc), np.std(all_acc)])
    histories[name] = (tr_l, te_l, te_a)

# ------------------------------------------------------------
# 9. Results table
# ------------------------------------------------------------
results_df = pd.DataFrame(
    results, columns=["Model", "Mean Accuracy", "Std"]
)

print("\n=== Final Performance Table ===")
print(results_df)

# ------------------------------------------------------------
# 10. Plot curves
# ------------------------------------------------------------
plt.figure(figsize=(16,5))

# Loss
plt.subplot(1,2,1)
for name, (tr_l, te_l, _) in histories.items():
    plt.plot(te_l, label=name)
plt.title("Test Loss per Epoch")
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.legend()

# Accuracy
plt.subplot(1,2,2)
for name, (_, _, te_a) in histories.items():
    plt.plot(te_a, label=name)
plt.title("Test Accuracy per Epoch")
plt.xlabel("Epoch")
plt.ylabel("Accuracy")
plt.legend()

plt.tight_layout()
plt.show()

# ------------------------------------------------------------
# 11. Confusion Matrix (AHP model)
# ------------------------------------------------------------
best_model = AttentionClassifier(ahp=True)
best_model.eval()
with torch.no_grad():
    preds = best_model(pad_batch(encoded_texts.tolist())).argmax(1)

cm = confusion_matrix(y, preds)

plt.figure(figsize=(10,8))
sns.heatmap(cm, xticklabels=le.classes_, yticklabels=le.classes_,
            cmap="Blues", annot=False)
plt.title("Confusion Matrix – AHP-Enhanced Attention")
plt.xlabel("Predicted")
plt.ylabel("True")
plt.tight_layout()
plt.show()

# ------------------------------------------------------------
# 12. ROC Curve (Micro-average)
# ------------------------------------------------------------
y_bin = F.one_hot(torch.tensor(y), num_classes=num_classes)
probs = F.softmax(best_model(pad_batch(encoded_texts.tolist())), dim=1)

fpr, tpr, _ = roc_curve(y_bin.ravel(), probs.detach().numpy().ravel())
auc = roc_auc_score(y_bin, probs.detach().numpy(), average="micro")

plt.figure(figsize=(7,7))
plt.plot(fpr, tpr, label=f"AHP-Attention (AUC={auc:.2f})", linewidth=3)
plt.plot([0,1], [0,1], "--", color="gray")
plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.title("Micro-average ROC Curve")
plt.legend()
plt.show()
```

```python
import random
import pandas as pd
import numpy as np

random.seed(42)
np.random.seed(42)

# ===============================
# Symptom Space (WHO-inspired)
# ===============================
SYMPTOMS = {
    "location": ["ankle", "leg", "foot", "face", "trunk"],
    "lesion_number": ["single", "few", "multiple"],
    "lesion_type": ["ulcer", "nodule", "swelling", "itchy"],
    "base": ["clean", "crusted", "necrotic"],
    "bilateral": ["yes", "no"],
    "pain": ["yes", "no"],
    "edema": ["yes", "no"],
    "varicose_veins": ["yes", "no"]
}

# ===============================
# Clinical Categories (NOT fine diseases)
# ===============================
CLASSES = {
    "Ulcerative": {
        "lesion_type": ["ulcer"],
        "location": ["ankle", "leg"],
        "bilateral": ["yes", "no"],
        "base": ["crusted", "necrotic"]
    },
    "Nodular": {
        "lesion_type": ["nodule"],
        "location": ["face", "trunk"],
        "pain": ["no"]
    },
    "Swelling": {
        "lesion_type": ["swelling"],
        "edema": ["yes"],
        "bilateral": ["yes"]
    },
    "Parasitic_Itchy": {
        "lesion_type": ["itchy"],
        "pain": ["yes"]
    }
}

# ===============================
# Generate one patient
# ===============================
def generate_patient(label, noise_prob=0.15):
    profile = {}
    rules = CLASSES[label]

    for symptom, values in SYMPTOMS.items():
        if symptom in rules and random.random() > noise_prob:
            profile[symptom] = random.choice(rules[symptom])
        else:
            profile[symptom] = random.choice(values)

    profile["label"] = label
    return profile

# ===============================
# Generate dataset
# ===============================
def generate_dataset(n_samples=600):
    data = []
    labels = list(CLASSES.keys())
    for _ in range(n_samples):
        label = random.choice(labels)
        data.append(generate_patient(label))
    return pd.DataFrame(data)

df = generate_dataset(650)
df.to_csv("synthetic_ntds_dataset.csv", index=False)
print(df.head())
```

```python
import torch
import torch.nn as nn
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.preprocessing import LabelEncoder
from sklearn.model_selection import StratifiedKFold
from sklearn.metrics import accuracy_score
import os

# ===============================
# Load dataset
# ===============================
df = pd.read_csv("synthetic_ntds_dataset.csv")

X = df.drop(columns=["label"])
y = LabelEncoder().fit_transform(df["label"])

X = pd.get_dummies(X)
X = torch.tensor(X.values, dtype=torch.float32)
y = torch.tensor(y, dtype=torch.long)

INPUT_DIM = X.shape[1]
N_CLASSES = len(np.unique(y))

# ===============================
# AHP weights (clinical importance)
# ===============================
AHP_WEIGHTS = torch.tensor(
    np.random.uniform(0.7, 1.3, INPUT_DIM),
    dtype=torch.float32
)

# ===============================
# Models
# ===============================
class BaselineAttention(nn.Module):
    def __init__(self):
        super().__init__()
        self.att = nn.Linear(INPUT_DIM, INPUT_DIM)
        self.fc = nn.Linear(INPUT_DIM, N_CLASSES)

    def forward(self, x):
        w = torch.softmax(self.att(x), dim=1)
        x = w * x
        return self.fc(x), w

class AHPTransformer(nn.Module):
    def __init__(self):
        super().__init__()
        self.att = nn.Linear(INPUT_DIM, INPUT_DIM)
        self.fc = nn.Linear(INPUT_DIM, N_CLASSES)

    def forward(self, x):
        scores = self.att(x)
        scores = scores * AHP_WEIGHTS
        w = torch.softmax(scores, dim=1)
        x = w * x
        return self.fc(x), w

# ===============================
# Training & stability evaluation
# ===============================
def evaluate(model_class, noise_std=0.0):
    accs = []
    skf = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

    for train, test in skf.split(X, y):
        model = model_class()
        opt = torch.optim.Adam(model.parameters(), lr=0.01)
        loss_fn = nn.CrossEntropyLoss()

        Xtr, Xte = X[train], X[test]
        ytr, yte = y[train], y[test]

        if noise_std > 0:
            Xtr = Xtr + noise_std * torch.randn_like(Xtr)

        for _ in range(40):
            opt.zero_grad()
            out, _ = model(Xtr)
            loss = loss_fn(out, ytr)
            loss.backward()
            opt.step()

        preds = torch.argmax(model(Xte)[0], dim=1)
        accs.append(accuracy_score(yte, preds))

    return np.mean(accs), np.std(accs)

results = {
    "Baseline Attention": evaluate(BaselineAttention),
    "AHP-Transformer": evaluate(AHPTransformer),
    "AHP-Transformer + Noise": evaluate(AHPTransformer, noise_std=0.2)
}

df_results = pd.DataFrame([
    {"Model": k, "Mean Accuracy": v[0], "Std (Stability)": v[1]}
    for k, v in results.items()
])

os.makedirs("figures", exist_ok=True)
df_results.to_csv("figures/results_table.csv", index=False)

# ===============================
# Plot Stability
# ===============================
plt.figure(figsize=(8,5))
plt.bar(df_results["Model"], df_results["Mean Accuracy"],
        yerr=df_results["Std (Stability)"], capsize=6)
plt.title("Model Stability under Data Scarcity")
plt.ylabel("Accuracy (mean ± std)")
plt.tight_layout()
plt.savefig("figures/stability_comparison.png", dpi=300)
plt.close()

print(df_results)
```

```python
# ============================================================
# AHP-Enhanced Attention for NTDs
# Comprehensive Experimental Evaluation & Visualization
# ============================================================

import os
import time
import numpy as np
import pandas as pd
import torch
import torch.nn as nn
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.preprocessing import LabelEncoder
from sklearn.model_selection import StratifiedKFold
from sklearn.metrics import accuracy_score, confusion_matrix
from sklearn.manifold import TSNE

# ============================================================
# Global plotting configuration (ARTICLE QUALITY)
# ============================================================
plt.rcParams.update({
    "font.size": 16,
    "axes.titlesize": 18,
    "axes.labelsize": 16,
    "legend.fontsize": 14,
    "xtick.labelsize": 14,
    "ytick.labelsize": 14,
    "lines.linewidth": 3
})

sns.set_style("whitegrid")

FIG_DIR = "figures"
os.makedirs(FIG_DIR, exist_ok=True)

# ============================================================
# 1. LOAD DATASET
# ============================================================
df = pd.read_csv("synthetic_ntds_dataset.csv")

X = pd.get_dummies(df.drop(columns=["label"]))
y = LabelEncoder().fit_transform(df["label"])
class_names = np.unique(df["label"])

X = torch.tensor(X.values, dtype=torch.float32)
y = torch.tensor(y, dtype=torch.long)

INPUT_DIM = X.shape[1]
N_CLASSES = len(class_names)

# ============================================================
# 2. AHP WEIGHTS (CLINICAL PRIORITY)
# ============================================================
np.random.seed(42)
AHP_WEIGHTS = torch.tensor(
    np.random.uniform(0.8, 1.2, INPUT_DIM),
    dtype=torch.float32
)

# ============================================================
# 3. MODELS
# ============================================================
class BaselineAttention(nn.Module):
    def __init__(self):
        super().__init__()
        self.att = nn.Linear(INPUT_DIM, INPUT_DIM)
        self.fc = nn.Linear(INPUT_DIM, N_CLASSES)

    def forward(self, x):
        weights = torch.softmax(self.att(x), dim=1)
        x = weights * x
        return self.fc(x), weights


class AHPAttention(nn.Module):
    def __init__(self):
        super().__init__()
        self.att = nn.Linear(INPUT_DIM, INPUT_DIM)
        self.fc = nn.Linear(INPUT_DIM, N_CLASSES)

    def forward(self, x):
        scores = self.att(x) * AHP_WEIGHTS
        weights = torch.softmax(scores, dim=1)
        x = weights * x
        return self.fc(x), weights


class RandomWeightedAttention(nn.Module):
    def __init__(self):
        super().__init__()
        self.rand_w = torch.rand(INPUT_DIM)
        self.att = nn.Linear(INPUT_DIM, INPUT_DIM)
        self.fc = nn.Linear(INPUT_DIM, N_CLASSES)

    def forward(self, x):
        scores = self.att(x) * self.rand_w
        weights = torch.softmax(scores, dim=1)
        x = weights * x
        return self.fc(x), weights


MODELS = {
    "Baseline Attention": BaselineAttention,
    "AHP-Enhanced Attention": AHPAttention,
    "Random-Weighted Attention": RandomWeightedAttention
}

# ============================================================
# 4. TRAIN & EVALUATE
# ============================================================
def train_and_evaluate(model_cls, noise_std=0.0):
    skf = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

    accs, losses, train_accs = [], [], []
    conf_total = np.zeros((N_CLASSES, N_CLASSES))
    embeddings_all, labels_all = [], []

    inference_times = []

    for train_idx, test_idx in skf.split(X, y):
        model = model_cls()
        optimizer = torch.optim.Adam(model.parameters(), lr=0.01)
        loss_fn = nn.CrossEntropyLoss()

        Xtr, Xte = X[train_idx], X[test_idx]
        ytr, yte = y[train_idx], y[test_idx]

        if noise_std > 0:
            Xtr = Xtr + noise_std * torch.randn_like(Xtr)

        fold_losses, fold_accs = [], []

        for epoch in range(40):
            optimizer.zero_grad()
            out, att = model(Xtr)
            loss = loss_fn(out, ytr)
            loss.backward()
            optimizer.step()

            preds = torch.argmax(out, dim=1)
            fold_losses.append(loss.item())
            fold_accs.append(accuracy_score(ytr, preds))

        # --- Evaluation
        start = time.time()
        out_test, att_test = model(Xte)
        inference_times.append(time.time() - start)

        preds_test = torch.argmax(out_test, dim=1)
        accs.append(accuracy_score(yte, preds_test))
        conf_total += confusion_matrix(yte, preds_test)

        embeddings_all.append(att_test.detach().numpy())
        labels_all.append(yte.numpy())

        losses.append(fold_losses)
        train_accs.append(fold_accs)

    return {
        "acc_mean": np.mean(accs),
        "acc_std": np.std(accs),
        "loss_curve": np.mean(losses, axis=0),
        "train_curve": np.mean(train_accs, axis=0),
        "confusion": conf_total,
        "embeddings": np.vstack(embeddings_all),
        "labels": np.concatenate(labels_all),
        "inference_time": np.mean(inference_times)
    }

# ============================================================
# 5. RUN EXPERIMENTS
# ============================================================
results = {}
for name, model in MODELS.items():
    results[name] = train_and_evaluate(model)

# ============================================================
# 6. STABILITY TABLE
# ============================================================
df_results = pd.DataFrame([
    {
        "Model": k,
        "Mean Accuracy": v["acc_mean"],
        "Std (Stability)": v["acc_std"],
        "Inference Time (s)": v["inference_time"]
    }
    for k, v in results.items()
])

df_results.to_csv(f"{FIG_DIR}/results_table.csv", index=False)

# ============================================================
# 7. LOSS CURVES (SAME FIGURE)
# ============================================================
plt.figure(figsize=(9,6))
for name, res in results.items():
    plt.plot(res["loss_curve"], label=f"{name} – Loss")
    

plt.xlabel("Epochs")
plt.ylabel("Loss ")
plt.title("Training Dynamics Comparison")
plt.legend()
plt.tight_layout()
plt.savefig(f"{FIG_DIR}/loss_curves.pdf")
plt.savefig(f"{FIG_DIR}/loss_curves.svg")
plt.close()
# ============================================================
# 7'.  TRAINING CURVES (SAME FIGURE)
# ============================================================
plt.figure(figsize=(9,6))
for name, res in results.items():
    
    plt.plot(res["train_curve"], linestyle="--", label=f"{name} – Train Acc")

plt.xlabel("Epochs")
plt.ylabel("Accuracy")
plt.title("Training Dynamics Comparison")
plt.legend()
plt.tight_layout()
plt.savefig(f"{FIG_DIR}/training_curves.pdf")
plt.savefig(f"{FIG_DIR}/training_curves.svg")
plt.close()
# ============================================================
# 8. CONFUSION MATRICES (SUBPLOTS)
# ============================================================
fig, axes = plt.subplots(1, 3, figsize=(18,6))

for ax, (name, res) in zip(axes, results.items()):
    sns.heatmap(
        res["confusion"],
        annot=True,
        fmt=".0f",
        cmap="Blues",
        xticklabels=class_names,
        yticklabels=class_names,
        ax=ax
    )
    ax.set_title(name)
    ax.set_xlabel("Predicted")
    ax.set_ylabel("True")

plt.tight_layout()
plt.savefig(f"{FIG_DIR}/confusion_matrices.pdf")
plt.savefig(f"{FIG_DIR}/confusion_matrices.svg")
plt.close()

# ============================================================
# 9. t-SNE VISUALIZATION (SUBPLOTS)
# ============================================================
fig, axes = plt.subplots(1, 3, figsize=(18,6))

for ax, (name, res) in zip(axes, results.items()):
    tsne = TSNE(n_components=2, perplexity=25, random_state=42)
    emb_2d = tsne.fit_transform(res["embeddings"])

    scatter = ax.scatter(
        emb_2d[:,0],
        emb_2d[:,1],
        c=res["labels"],
        cmap="tab10",
        alpha=0.7
    )
    ax.set_title(name)
    ax.set_xlabel("t-SNE 1")
    ax.set_ylabel("t-SNE 2")

plt.tight_layout()
plt.savefig(f"{FIG_DIR}/tsne_comparison.pdf")
plt.savefig(f"{FIG_DIR}/tsne_comparison.svg")
plt.close()

# ============================================================
# 10. INFERENCE TIME BARPLOT
# ============================================================
plt.figure(figsize=(8,5))
plt.bar(
    df_results["Model"],
    df_results["Inference Time (s)"]
)
plt.ylabel("Inference Time (seconds)")
plt.title("Inference Time Comparison")
plt.xticks(rotation=15)
plt.tight_layout()
plt.savefig(f"{FIG_DIR}/inference_time.pdf")
plt.savefig(f"{FIG_DIR}/inference_time.svg")
plt.close()

# ============================================================
# 11. STABILITY BARPLOT
# ============================================================
plt.figure(figsize=(8,5))
plt.bar(
    df_results["Model"],
    df_results["Mean Accuracy"],
    yerr=df_results["Std (Stability)"],
    capsize=6
)
plt.ylabel("Accuracy (mean ± std)")
plt.title("Model Stability under Data Scarcity")
plt.xticks(rotation=15)
plt.tight_layout()
plt.savefig(f"{FIG_DIR}/stability.pdf")
plt.savefig(f"{FIG_DIR}/stability.svg")
plt.close()

print("✅ All experiments completed. Figures saved in 'figures/'")
print(df_results)
```

```python
import torch
import torch.nn as nn
import pandas as pd
import numpy as np
from sklearn.preprocessing import LabelEncoder
from sklearn.metrics import accuracy_score # Pour une vérification rapide de la précision
import pickle
import os

# --- Configuration ---
MODEL_DIR = "models"
os.makedirs(MODEL_DIR, exist_ok=True) # Crée le dossier 'models' s'il n'existe pas
DATA_PATH = "synthetic_ntds_dataset.csv" # Assurez-vous que ce fichier est dans le même répertoire

# ===============================
# 1. Chargement du jeu de données
# ===============================
try:
    df = pd.read_csv(DATA_PATH)
    print(f"Dataset '{DATA_PATH}' chargé avec succès.")
except FileNotFoundError:
    print(f"Erreur: Le fichier de données '{DATA_PATH}' est introuvable.")
    print("Création d'un jeu de données factice pour la démonstration.")
    # Création d'un jeu de données factice si le fichier n'est pas trouvé
    data = {
        "location": np.random.choice(["ankle", "leg", "foot", "face", "trunk"], 200),
        "lesion_number": np.random.choice(["single", "few", "multiple"], 200),
        "lesion_type": np.random.choice(["ulcer", "nodule", "swelling", "itchy"], 200),
        "base": np.random.choice(["clean", "crusted", "necrotic"], 200),
        "bilateral": np.random.choice(["yes", "no"], 200),
        "pain": np.random.choice(["yes", "no"], 200),
        "edema": np.random.choice(["yes", "no"], 200),
        "varicose_veins": np.random.choice(["yes", "no"], 200),
        "label": np.random.choice(["Ulcerative", "Nodular", "Swelling", "Itchy"], 200)
    }
    df = pd.DataFrame(data)
    df.to_csv(DATA_PATH, index=False)
    print(f"Jeu de données factice sauvegardé sous '{DATA_PATH}'.")

# ===============================
# 2. Prétraitement des données et sauvegarde des préprocesseurs
# ===============================
X = df.drop(columns=["label"])
y_raw = df["label"]

# 2.1. Encoder les étiquettes cibles (y) et sauvegarder le LabelEncoder
label_encoder = LabelEncoder()
y = label_encoder.fit_transform(y_raw)
with open(os.path.join(MODEL_DIR, "label_encoder.pkl"), "wb") as f:
    pickle.dump(label_encoder, f)
print(f"LabelEncoder sauvegardé dans {os.path.join(MODEL_DIR, 'label_encoder.pkl')}")

# 2.2. Appliquer l'encodage one-hot aux caractéristiques (X) et sauvegarder les noms de colonnes
X_processed = pd.get_dummies(X)
feature_columns = X_processed.columns.tolist()
with open(os.path.join(MODEL_DIR, "feature_columns.pkl"), "wb") as f:
    pickle.dump(feature_columns, f)
print(f"Noms des colonnes des caractéristiques sauvegardés dans {os.path.join(MODEL_DIR, 'feature_columns.pkl')}")

# Convertir en tenseurs PyTorch
X_tensor = torch.tensor(X_processed.values, dtype=torch.float32)
y_tensor = torch.tensor(y, dtype=torch.long)

INPUT_DIM = X_tensor.shape[1]
N_CLASSES = len(label_encoder.classes_) # Nombre de classes à partir de l'encodeur

print(f"Dimension d'entrée (INPUT_DIM): {INPUT_DIM}")
print(f"Nombre de classes (N_CLASSES): {N_CLASSES}")

# ===============================
# 3. Poids AHP (importance clinique)
# ===============================
# Ces poids sont générés aléatoirement pour l'exemple. Dans un scénario réel,
# ils proviendraient de l'expertise clinique via une méthode AHP.
AHP_WEIGHTS = torch.tensor(
    np.random.uniform(0.7, 1.3, INPUT_DIM),
    dtype=torch.float32
)
# Sauvegarder les poids AHP
torch.save(AHP_WEIGHTS, os.path.join(MODEL_DIR, "ahp_weights.pt"))
print(f"Poids AHP sauvegardés dans {os.path.join(MODEL_DIR, 'ahp_weights.pt')}")

# ===============================
# 4. Définition du Modèle AHPTransformer
# (Doit être identique à celle utilisée dans l'application Streamlit)
# ===============================
class AHPTransformer(nn.Module):
    def __init__(self, input_dim, n_classes, ahp_weights):
        super().__init__()
        self.att = nn.Linear(input_dim, input_dim)
        self.fc = nn.Linear(input_dim, n_classes)
        # Enregistrer ahp_weights comme un buffer pour qu'il soit sauvegardé avec le modèle
        # mais ne soit pas un paramètre entraînable.
        self.register_buffer('ahp_weights', ahp_weights)

    def forward(self, x):
        scores = self.att(x)
        scores = scores * self.ahp_weights # Appliquer les poids AHP
        w = torch.softmax(scores, dim=1) # Calculer les poids d'attention
        x = w * x # Appliquer l'attention pondérée
        return self.fc(x), w

# ===============================
# 5. Entraînement du modèle AHPTransformer pour le déploiement
# ===============================
print("\nDébut de l'entraînement du modèle AHP-Transformer...")
model = AHPTransformer(INPUT_DIM, N_CLASSES, AHP_WEIGHTS)
opt = torch.optim.Adam(model.parameters(), lr=0.01)
loss_fn = nn.CrossEntropyLoss()

NUM_EPOCHS = 150 # Nombre d'époques augmenté pour un entraînement sur le jeu de données complet

for epoch in range(NUM_EPOCHS):
    opt.zero_grad()
    out, _ = model(X_tensor)
    loss = loss_fn(out, y_tensor)
    loss.backward()
    opt.step()
    if (epoch + 1) % 25 == 0:
        print(f"Époque {epoch+1}/{NUM_EPOCHS}, Perte: {loss.item():.4f}")

# Évaluation rapide sur les données d'entraînement (pour vérification)
model.eval() # Passer le modèle en mode évaluation
with torch.no_grad(): # Désactiver le calcul des gradients
    preds = torch.argmax(model(X_tensor)[0], dim=1)
    train_accuracy = accuracy_score(y_tensor.numpy(), preds.numpy())
    print(f"Précision d'entraînement finale: {train_accuracy:.4f}")

# ===============================
# 6. Sauvegarde du modèle entraîné
# ===============================
model_path = os.path.join(MODEL_DIR, "ahp_transformer_model.pth")
torch.save(model.state_dict(), model_path)
print(f"Modèle AHP-Transformer entraîné sauvegardé dans {model_path}")

print("\nConfiguration du backend terminée. Modèle et préprocesseurs sauvegardés.")
print("Vous pouvez maintenant exécuter l'application Streamlit.")
```
